# Decorators

```python
def func():
  print('hello')
a=func()
print(a)
del func                 #1st class function
func()
```

```python
def my_decorator(func):
  def wrapper():
    print('###############')
    func()
    print('###############')
  return wrapper

def hello():
  print('hello')

a= my_decorator(hello)
a()
```

```python
def my_decorator(func):
  def wrapper():
    print('###############')
    func()
    print('###############')
  return wrapper

@my_decorator
def hello():
  print('hello')
hello()
```

```python
import time
def timer(func):
  def wrapper(*args):
    start=time.time()
    func(*args)
    print("Time taken by",func.__name__,time.time()-start,'secs')
  return wrapper
@timer
def hello():
  print('hello world')
  time.sleep(2)
@timer
def display():
  print('display something')
  time.sleep(4)

hello()
display()
```

```python
def sanity_check(data_type):
  def outer_wrapper(func):
    def inner_wrapper(*args):
      if type(args[0])==data_type:
        func(*args)
      else:
        raise TypeError('Incorrect datatype')
    return inner_wrapper
  return outer_wrapper

@sanity_check(int)
def square(num):
  print(num**2)

square(2)
square('hahahaha')
```
