# Namespace


```python
#local & global
a=2
def temp():
  b=3
  print(b)
temp()
print(a)
```

```python
a=2
def temp():
  a+=1
  print(a)
temp()
print(a)
```

```python
a=2
def temp():
  global a
  a+=1
  print(a)
temp()
print(a)  #not good programming practice, can cause problems in rest functions
```

```python
#built-in scope
print('hello')
import builtins
print(dir(builtins))
```

```python
#enclosing scope
def outer():
  a=3
  def inner():          #LEGB rule
    a=4
    print(a)
  inner()
  print('Outer func')
a=1
outer()
print('main program')
```

```python
def outer():
  a=3
  def inner():
    nonlocal a
    a+=1
    print('inner',a)
  inner()
  print('Outer func',a)
a=1
outer()
print('main program')
```
